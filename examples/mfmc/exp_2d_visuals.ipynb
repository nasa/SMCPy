```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform
from exp_2d import M_HF, M_LF, generate_noisy_data, x, resolution, M_LF3
```

```python
THETA = np.array([[1/20,1]])
# Evaluate the function over the grid
Z_HF = M_HF(THETA)
# Evaluate the function over the grid
Z_LF = M_LF(THETA)
noise_st_dev = 0.08
meas_data = generate_noisy_data(THETA, noise_st_dev)
```

```python
# Create a 1 by 3 subplot figure
fig, axes = plt.subplots(2, 2, figsize=(8, 9))
axes = axes.flatten()
size = 3
axes[0].scatter(x, Z_HF, s = size)
axes[0].set_title('HiFi Model')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].grid()

axes[1].scatter(x, Z_LF, s = size)
axes[1].set_title('LoFi Model')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].grid()

axes[2].scatter(x, Z_HF - Z_LF, s = size)
axes[2].set_title('Difference')
axes[2].set_xlabel('x')
axes[2].set_ylabel('y')
axes[2].grid()

axes[3].scatter(x, meas_data, s = size)
axes[3].set_title('Observed Data')
axes[3].set_xlabel('x')
axes[3].set_ylabel('y')
axes[3].grid()

# Adjust layout so subplots don't overlap
plt.tight_layout()

# Display the plot
plt.show()
```
